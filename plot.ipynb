```python
# 1. Ladda data (ersätt 'data.csv' med din fil)
df = pd.read_csv('diabetic_data.csv')

# Lista på alla medicinkolumner (från din header)
medication_cols = [
    'metformin', 'repaglinide', 'nateglinide', 'chlorpropamide', 'glimepiride', 
    'acetohexamide', 'glipizide', 'glyburide', 'tolbutamide', 'pioglitazone', 
    'rosiglitazone', 'acarbose', 'miglitol', 'troglitazone', 'tolazamide', 
    'examide', 'citoglipton', 'insulin', 'glyburide-metformin', 'glipizide-metformin', 
    'glimepiride-pioglitazone', 'metformin-rosiglitazone', 'metformin-pioglitazone'
]

# Vi vill se hur medicinen korrelerar med om man blev återinskriven (readmitted)
target = 'readmitted'

# Loopa igenom varje medicin och skapa ett stapeldiagram
for med in medication_cols:
    # Kontrollera om medicinen faktiskt har mer än ett värde (vissa kan vara 'No' för alla)
    if df[med].nunique() > 1:
        plt.figure(figsize=(8, 5))
        
        # Skapa en korstabell mellan medicinen och återinskrivning
        ct = pd.crosstab(df[med], df[target], normalize='index')
        
        # Plotta som ett staplat stapeldiagram
        ct.plot(kind='bar', stacked=True, alpha=0.8)
        
        plt.title(f"Inverkan av {med} på {target}")
        plt.xlabel(f"Dosering/Status för {med}")
        plt.ylabel("Andel patienter")
        plt.legend(title="Readmitted", loc='upper right')
        plt.xticks(rotation=0)
        plt.show()
    else:
        print(f"Hoppar över {med}: Innehåller endast ett värde ({df[med].unique()[0]})")
```
